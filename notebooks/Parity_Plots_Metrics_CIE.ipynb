```python
import pickle
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import mean_absolute_error, r2_score
from mpl_toolkits.mplot3d import Axes3D as ax
```

```python
def retrieve_points(path):
    """
    Load a pickled matplotlib figure containing a scatter plot,
    and return the x and y coordinates as a dictionary.

    Returns:
        list: A list of tuples, each containing the x and y coordinates of a point.
    """

    with open(path, "rb") as f:
        fig = pickle.load(f)

    ax = fig.axes[0]

    scatter = ax.collections[0]
    offsets = scatter.get_offsets()

    x = offsets[:, 0]
    y = offsets[:, 1]

    return {'x': x, 'y': y}


def load_parity(path):
  with open(path, "rb") as f:
    fig = pickle.load(f)

  fig.show()


def mape(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

'''
def load_model_performance(path):

    points = retrieve_points(path)

    trues = points['x']
    preds = points['y']

    rmse = np.sqrt(np.mean((trues - preds)**2))
    print("RMSE:", rmse)

    mae = mean_squared_error(trues, preds)
    print("MAE:", mae)

    r2 = r2_score(trues, preds)
    print("R²:", r2)

    print("MAPE:", mape(trues, preds))

    print('')

    x_loc = plt.xlim()[1] - (plt.xlim()[1] - plt.xlim()[0]) * 0.2
    y_loc = plt.ylim()[0] + (plt.ylim()[1] - plt.ylim()[0]) * 0.2

    plt.text(x_loc, y_loc, "Your text", fontsize=12, color='red')
    plt.show()
    '''

def load_model_performance(path):

    points = retrieve_points(path)

    trues = points['x']
    preds = points['y']

    # --- Compute metrics ---
    rmse = np.sqrt(np.mean((trues - preds) ** 2))
    mae = mean_absolute_error(trues, preds)
    r2 = r2_score(trues, preds)
    mape_val = mape(trues, preds)

    # --- Get current axes ---
    ax = plt.gca()

    # --- Build metrics text ---
    metrics_text = (
        f"RMSE: {rmse:.6g}\n"
        f"MAE:  {mae:.6g}\n"
        f"R²:   {r2:.6g}\n"
        f"MAPE: {mape_val:.6g}%"
    )

    # --- Add text box to bottom-right ---
    ax.text(
        0.98, 0.02, metrics_text,
        transform=ax.transAxes,
        fontsize=10,
        ha='right',
        va='bottom',
        bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.8)
    )

    plt.show()
```
