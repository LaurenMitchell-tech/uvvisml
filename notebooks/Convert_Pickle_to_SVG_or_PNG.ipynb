```python
import pickle
import os
from google.colab import files
import zipfile

uploaded = files.upload()
```

```python
def pkl_to_svg(paths):

    svgs = []

    for path in paths:

        with open(path, "rb") as f:
            fig = pickle.load(f)

        svg_path = os.path.splitext(path)[0] + ".svg"
        fig.savefig(f'/content/{svg_path}', bbox_inches="tight")
        svgs.append(svg_path)

    for file in svgs:
        files.download(file)
```

```python
def pkl_to_png(paths):

    pngs = []

    for path in paths:

        with open(path, "rb") as f:
            fig = pickle.load(f)

        png_path = os.path.splitext(path)[0] + ".png"
        fig.savefig(f'/content/{png_path}', bbox_inches="tight")
        pngs.append(png_path)

    for file in pngs:
        files.download(file)
```

```python
def pkl_to_svg_zip(paths, zip_name="plots.zip"):
    svg_files = []

    # Save all SVGs in Colab VM
    for path in paths:
        with open(path, "rb") as f:
            fig = pickle.load(f)
        svg_path = os.path.splitext(path)[0] + ".svg"
        fig.savefig(svg_path, bbox_inches="tight")
        svg_files.append(svg_path)

    # Create a ZIP containing all SVGs
    with zipfile.ZipFile(zip_name, "w") as zipf:
        for file in svg_files:
            zipf.write(file)

    # Download the ZIP
    files.download(zip_name)
```

```python
def pkl_to_png_zip(paths, zip_name="plots.zip"):
    png_files = []

    # Save all PNGs in Colab VM
    for path in paths:
        with open(path, "rb") as f:
            fig = pickle.load(f)
        png_path = os.path.splitext(path)[0] + ".png"
        fig.savefig(png_path, bbox_inches="tight")
        png_files.append(png_path)

    # Create a ZIP containing all PNGs
    with zipfile.ZipFile(zip_name, "w") as zipf:
        for file in png_files:
            zipf.write(file)

    # Download the ZIP
    files.download(zip_name)
```
